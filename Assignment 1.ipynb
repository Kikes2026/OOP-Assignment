```python
class Student:
    university = "Kabarak University"

    def __init__(self, name, student_id, programme):
        self.name = name               
        self.student_id = student_id
        self.programme = programme

    def display_details(self):
        print("----- Student Details -----")
        print(f"Name:       {self.name}")
        print(f"Student ID: {self.student_id}")
        print(f"Programme:  {self.programme}")
        print(f"University: {Student.university}")
        print("---------------------------")

    def change_programme(self, new_programme):
        old = self.programme
        self.programme = new_programme
        print(f"[{self.name}] Programme changed: '{old}' -> '{new_programme}'")

    @classmethod
    def change_university(cls, new_university):
        old = cls.university
        cls.university = new_university
        print(f"[CLASS] University changed: '{old}' -> '{new_university}'")

    @staticmethod
    def is_valid_id(student_id):
        return len(str(student_id)) >= 5

print("STUDENT ID VALIDATION")
print(f"Is BDSA/ON/2026 valid? {Student.is_valid_id('BDSA/2012')}")
print(f"Is bds valid? {Student.is_valid_id('bds')}")

assert Student.is_valid_id("BDSA/ON/2026") is True
assert Student.is_valid_id("bds") is False
print("\nStudent ID validation tests passed.")
```
